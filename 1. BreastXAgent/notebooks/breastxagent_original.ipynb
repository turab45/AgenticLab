```python
# ============================================================
# CELL 2: Run MONAI Breast Density Classification WITHOUT MONAI
# Hugging Face repo: MONAI/breast_density_classification
# Task: classify breast density A/B/C/D on sample_data images
# ============================================================

import os
import glob
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from torch import nn
from huggingface_hub import snapshot_download
from torchvision.models import inception_v3
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Torch version:", torch.__version__)

try:
    import torchvision
    print("TorchVision version:", torchvision.__version__)
except Exception as e:
    print("TorchVision import issue:", e)

# -----------------------------
# 1. Download Hugging Face repo
# -----------------------------
REPO_ID = "MONAI/breast_density_classification"
LOCAL_DIR = "/content/breast_density_classification"

snapshot_download(
    repo_id=REPO_ID,
    local_dir=LOCAL_DIR,
    local_dir_use_symlinks=False
)

print("Repo downloaded to:", LOCAL_DIR)

# -----------------------------
# 2. Class labels
# -----------------------------
CLASS_NAMES = ["A", "B", "C", "D"]

CLASS_DESCRIPTIONS = {
    "A": "Almost entirely fatty",
    "B": "Scattered fibroglandular density",
    "C": "Heterogeneously dense",
    "D": "Extremely dense"
}

# -----------------------------
# 3. Device
# CPU is enough for sample images
# -----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# -----------------------------
# 4. Recreate the MONAI TorchVisionFCModel using pure torchvision
# Model: Inception-v3
# Output classes: 4
# Input size: 299 x 299
# -----------------------------
class InceptionMONAIStyleAdapter(nn.Module):
    def __init__(self, num_classes=4):
        super().__init__()

        base = inception_v3(
            weights=None,
            aux_logits=True,
            transform_input=False
        )

        # MONAI TorchVisionFCModel removes original classifier
        base.fc = nn.Identity()

        self.features = base
        self.fc = nn.Linear(2048, num_classes)

    def forward(self, x):
        x = self.features(x)

        if hasattr(x, "logits"):
            x = x.logits
        elif isinstance(x, (tuple, list)):
            x = x[0]

        x = torch.flatten(x, 1)
        x = self.fc(x)

        return x

model = InceptionMONAIStyleAdapter(num_classes=4).to(device)

# -----------------------------
# 5. Load checkpoint
# -----------------------------
model_path = os.path.join(LOCAL_DIR, "models", "model.pt")
print("Model checkpoint:", model_path)

try:
    checkpoint = torch.load(model_path, map_location=device, weights_only=True)
except Exception:
    checkpoint = torch.load(model_path, map_location=device, weights_only=False)

if isinstance(checkpoint, dict):
    if "model" in checkpoint:
        state_dict = checkpoint["model"]
    elif "state_dict" in checkpoint:
        state_dict = checkpoint["state_dict"]
    else:
        state_dict = checkpoint
else:
    state_dict = checkpoint

clean_state_dict = {}

for k, v in state_dict.items():
    new_k = k
    new_k = new_k.replace("module.", "")
    new_k = new_k.replace("network.", "")
    clean_state_dict[new_k] = v

print("\nExample checkpoint keys:")
for k in list(clean_state_dict.keys())[:10]:
    print(k)

missing, unexpected = model.load_state_dict(clean_state_dict, strict=False)

print("\nCheckpoint loaded.")
print("Missing keys:", len(missing))
print("Unexpected keys:", len(unexpected))

if len(missing) > 0:
    print("First missing keys:", missing[:10])

if len(unexpected) > 0:
    print("First unexpected keys:", unexpected[:10])

model.eval()

# -----------------------------
# 6. Preprocess image
# -----------------------------
def preprocess_image(image_path):
    img = Image.open(image_path).convert("RGB")
    img = img.resize((299, 299), Image.BILINEAR)

    arr = np.asarray(img).astype(np.float32)

    # Scale intensity to [0,1]
    arr_min = arr.min()
    arr_max = arr.max()
    arr = (arr - arr_min) / (arr_max - arr_min + 1e-8)

    # HWC -> CHW -> BCHW
    tensor = torch.from_numpy(arr).permute(2, 0, 1).unsqueeze(0)
    tensor = tensor.float()

    return tensor

# -----------------------------
# 7. Predict one image
# -----------------------------
def predict_image(image_path):
    x = preprocess_image(image_path).to(device)

    with torch.no_grad():
        logits = model(x)

        # Official MONAI bundle uses sigmoid-style output postprocessing
        sigmoid_scores = torch.sigmoid(logits)[0].detach().cpu().numpy()

        # Also compute softmax for reference
        softmax_scores = torch.softmax(logits, dim=1)[0].detach().cpu().numpy()

    pred_idx = int(np.argmax(sigmoid_scores))
    pred_class = CLASS_NAMES[pred_idx]
    confidence = float(sigmoid_scores[pred_idx])

    return pred_class, confidence, sigmoid_scores, softmax_scores

# -----------------------------
# 8. Find sample images
# -----------------------------
sample_dir = os.path.join(LOCAL_DIR, "sample_data")

image_paths = []

for ext in ["*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"]:
    image_paths.extend(glob.glob(os.path.join(sample_dir, "*", ext)))

image_paths = sorted(image_paths)

print("\nSample directory:", sample_dir)
print("Images found:", len(image_paths))

if len(image_paths) == 0:
    raise RuntimeError("No images found in sample_data/A, B, C, D folders.")

print("First image:", image_paths[0])

# -----------------------------
# 9. Run inference on all sample images
# -----------------------------
results = []

for img_path in image_paths:
    true_class = os.path.basename(os.path.dirname(img_path))

    pred_class, confidence, sigmoid_scores, softmax_scores = predict_image(img_path)

    results.append({
        "image": os.path.basename(img_path),
        "true_class": true_class,
        "predicted_class": pred_class,
        "predicted_description": CLASS_DESCRIPTIONS[pred_class],
        "confidence_sigmoid": confidence,
        "sigmoid_A": float(sigmoid_scores[0]),
        "sigmoid_B": float(sigmoid_scores[1]),
        "sigmoid_C": float(sigmoid_scores[2]),
        "sigmoid_D": float(sigmoid_scores[3]),
        "softmax_A": float(softmax_scores[0]),
        "softmax_B": float(softmax_scores[1]),
        "softmax_C": float(softmax_scores[2]),
        "softmax_D": float(softmax_scores[3]),
        "correct": true_class == pred_class,
        "path": img_path
    })

df = pd.DataFrame(results)
display(df)

csv_path = "/content/breast_density_predictions.csv"
df.to_csv(csv_path, index=False)

print("Saved predictions to:", csv_path)

# -----------------------------
# 10. Metrics
# -----------------------------
acc = accuracy_score(df["true_class"], df["predicted_class"])

print("\nAccuracy on sample images:", acc)

print("\nClassification report:")
print(
    classification_report(
        df["true_class"],
        df["predicted_class"],
        labels=CLASS_NAMES,
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

cm = confusion_matrix(
    df["true_class"],
    df["predicted_class"],
    labels=CLASS_NAMES
)

cm_df = pd.DataFrame(
    cm,
    index=[f"True_{c}" for c in CLASS_NAMES],
    columns=[f"Pred_{c}" for c in CLASS_NAMES]
)

print("\nConfusion matrix:")
display(cm_df)

# -----------------------------
# 11. Show sample predictions
# -----------------------------
def show_predictions(df, n=8):
    n = min(n, len(df))
    sample_df = df.head(n)

    cols = 4
    rows = int(np.ceil(n / cols))

    plt.figure(figsize=(16, 4 * rows))

    for i, row in enumerate(sample_df.itertuples(), 1):
        img = Image.open(row.path).convert("RGB")

        plt.subplot(rows, cols, i)
        plt.imshow(img, cmap="gray")
        plt.axis("off")

        title = (
            f"True: {row.true_class}\n"
            f"Pred: {row.predicted_class} | Conf: {row.confidence_sigmoid:.2f}"
        )

        plt.title(title)

    plt.tight_layout()
    plt.show()

show_predictions(df, n=8)
```

```python
# ============================================================
# CELL 3: Grad-CAM for Breast Density Classification Model
# Run this AFTER the working inference cell
# ============================================================

import os
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

# -----------------------------
# 1. Choose Grad-CAM target layer
# -----------------------------
# For Inception-v3, the last convolutional block is Mixed_7c.
# Our model wrapper is:
# model.features = inception_v3 backbone
# model.fc       = final 4-class classifier
target_layer = model.features.Mixed_7c

print("Using Grad-CAM target layer:", target_layer)


# -----------------------------
# 2. Grad-CAM class
# -----------------------------
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer

        self.activations = None
        self.gradients = None

        self.forward_hook = self.target_layer.register_forward_hook(
            self.save_activation
        )

        self.backward_hook = self.target_layer.register_full_backward_hook(
            self.save_gradient
        )

    def save_activation(self, module, input, output):
        self.activations = output.detach()

    def save_gradient(self, module, grad_input, grad_output):
        self.gradients = grad_output[0].detach()

    def remove_hooks(self):
        self.forward_hook.remove()
        self.backward_hook.remove()

    def generate(self, image_tensor, target_class_idx=None):
        self.model.eval()
        self.model.zero_grad()

        image_tensor = image_tensor.to(device)
        image_tensor.requires_grad_(True)

        logits = self.model(image_tensor)

        if target_class_idx is None:
            scores = torch.sigmoid(logits)
            target_class_idx = int(torch.argmax(scores, dim=1).item())

        target_score = logits[0, target_class_idx]
        target_score.backward(retain_graph=True)

        # Grad-CAM weights: global average pooling of gradients
        gradients = self.gradients[0]          # shape: [C, H, W]
        activations = self.activations[0]      # shape: [C, H, W]

        weights = torch.mean(gradients, dim=(1, 2))  # shape: [C]

        cam = torch.zeros(
            activations.shape[1:],
            dtype=torch.float32,
            device=activations.device
        )

        for i, w in enumerate(weights):
            cam += w * activations[i]

        cam = torch.relu(cam)

        # Normalize CAM to [0,1]
        cam = cam - cam.min()
        cam = cam / (cam.max() + 1e-8)

        cam = cam.detach().cpu().numpy()

        return cam, target_class_idx


# -----------------------------
# 3. Helper: overlay heatmap on image
# -----------------------------
def overlay_cam_on_image(image_path, cam, alpha=0.45):
    original = Image.open(image_path).convert("RGB")
    original_np = np.asarray(original).astype(np.float32) / 255.0

    # Resize CAM to original image size
    cam_img = Image.fromarray(np.uint8(cam * 255))
    cam_img = cam_img.resize(original.size, Image.BILINEAR)
    cam_resized = np.asarray(cam_img).astype(np.float32) / 255.0

    # Convert CAM to color heatmap
    heatmap = plt.get_cmap("jet")(cam_resized)[:, :, :3]

    overlay = (1 - alpha) * original_np + alpha * heatmap
    overlay = np.clip(overlay, 0, 1)

    return original_np, heatmap, overlay


# -----------------------------
# 4. Generate Grad-CAM for one image
# -----------------------------
gradcam = GradCAM(model, target_layer)

def generate_gradcam_for_image(image_path, save_dir="/content/gradcam_outputs"):
    os.makedirs(save_dir, exist_ok=True)

    # Preprocess image
    image_tensor = preprocess_image(image_path).to(device)

    # Forward once to get prediction
    model.eval()
    with torch.no_grad():
        logits = model(image_tensor)
        sigmoid_scores = torch.sigmoid(logits)[0].detach().cpu().numpy()

    pred_idx = int(np.argmax(sigmoid_scores))
    pred_class = CLASS_NAMES[pred_idx]
    confidence = float(sigmoid_scores[pred_idx])

    # Generate CAM using predicted class
    cam, target_idx = gradcam.generate(
        image_tensor=image_tensor,
        target_class_idx=pred_idx
    )

    original_np, heatmap, overlay = overlay_cam_on_image(image_path, cam)

    # Save Grad-CAM overlay
    base_name = os.path.splitext(os.path.basename(image_path))[0]
    save_path = os.path.join(
        save_dir,
        f"{base_name}_pred_{pred_class}_gradcam.png"
    )

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(original_np, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(heatmap)
    plt.title("Grad-CAM Heatmap")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(overlay)
    plt.title(f"Overlay\nPred: {pred_class}, Conf: {confidence:.2f}")
    plt.axis("off")

    plt.tight_layout()
    plt.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.close()

    return {
        "image_path": image_path,
        "predicted_class": pred_class,
        "confidence": confidence,
        "gradcam_path": save_path
    }


# -----------------------------
# 5. Generate Grad-CAM for all sample images
# -----------------------------
gradcam_results = []

for row in df.itertuples():
    result = generate_gradcam_for_image(row.path)
    result["true_class"] = row.true_class
    result["correct"] = row.true_class == result["predicted_class"]
    gradcam_results.append(result)

gradcam_df = pd.DataFrame(gradcam_results)

display(gradcam_df)

gradcam_csv_path = "/content/breast_density_gradcam_results.csv"
gradcam_df.to_csv(gradcam_csv_path, index=False)

print("Saved Grad-CAM results CSV:", gradcam_csv_path)
print("Grad-CAM images saved in: /content/gradcam_outputs")
```

```python
# ============================================================
# CELL 5: LangGraph Agentic Explanation Layer
# Uses gradcam_df generated from Cell 3
# No LLM yet: safe template-based report
# ============================================================

!pip -q install langgraph

from typing import TypedDict, Optional
from langgraph.graph import StateGraph, START, END
import pandas as pd
import os

# -----------------------------
# 1. State definition
# -----------------------------
class BreastDensityAgentState(TypedDict):
    image_path: str
    gradcam_path: str
    predicted_class: str
    confidence: float

    density_description: Optional[str]
    confidence_level: Optional[str]
    evidence_status: Optional[str]
    recommendation_level: Optional[str]
    final_report: Optional[str]


# -----------------------------
# 2. Class descriptions
# -----------------------------
DENSITY_DESCRIPTIONS = {
    "A": "almost entirely fatty breast tissue",
    "B": "scattered fibroglandular breast density",
    "C": "heterogeneously dense breast tissue",
    "D": "extremely dense breast tissue"
}


# -----------------------------
# 3. Node 1: interpret prediction
# -----------------------------
def interpret_prediction_node(state: BreastDensityAgentState):
    pred = state["predicted_class"]

    return {
        "density_description": DENSITY_DESCRIPTIONS.get(
            pred,
            "unknown breast density category"
        )
    }


# -----------------------------
# 4. Node 2: confidence analysis
# -----------------------------
def confidence_analysis_node(state: BreastDensityAgentState):
    conf = state["confidence"]

    if conf >= 0.85:
        confidence_level = "high"
    elif conf >= 0.70:
        confidence_level = "moderate"
    else:
        confidence_level = "low"

    return {
        "confidence_level": confidence_level
    }


# -----------------------------
# 5. Router: decide report type
# -----------------------------
def confidence_router(state: BreastDensityAgentState):
    if state["confidence_level"] == "low":
        return "uncertain"
    return "supported"


# -----------------------------
# 6. Node 3A: evidence check for supported cases
# -----------------------------
def evidence_supported_node(state: BreastDensityAgentState):
    if state["confidence_level"] == "high":
        evidence_status = "strongly supported by the classifier confidence and Grad-CAM visual evidence"
        recommendation_level = "standard evidence-guided report"
    else:
        evidence_status = "moderately supported; the Grad-CAM map should be reviewed with caution"
        recommendation_level = "cautious evidence-guided report"

    return {
        "evidence_status": evidence_status,
        "recommendation_level": recommendation_level
    }


# -----------------------------
# 7. Node 3B: uncertain case
# -----------------------------
def uncertain_node(state: BreastDensityAgentState):
    return {
        "evidence_status": "uncertain because the classifier confidence is low",
        "recommendation_level": "low-confidence warning"
    }


# -----------------------------
# 8. Node 4: final structured report
# -----------------------------
def final_report_node(state: BreastDensityAgentState):
    report = f"""
Breast Density Classification Report

Image:
{os.path.basename(state['image_path'])}

Predicted density class:
{state['predicted_class']}

Density interpretation:
The model predicts {state['density_description']}.

Confidence:
{state['confidence']:.3f} ({state['confidence_level']} confidence)

Visual evidence:
A Grad-CAM heatmap was generated to highlight image regions contributing to the model decision.
Grad-CAM file:
{state['gradcam_path']}

Evidence status:
{state['evidence_status']}

Agent decision:
{state['recommendation_level']}

Research note:
This output is generated by a lightweight agentic explanation layer around a pretrained breast-density classifier. It is intended for research and explanation support only, not clinical diagnosis.
""".strip()

    return {
        "final_report": report
    }


# -----------------------------
# 9. Build LangGraph
# -----------------------------
workflow = StateGraph(BreastDensityAgentState)

workflow.add_node("interpret_prediction", interpret_prediction_node)
workflow.add_node("confidence_analysis", confidence_analysis_node)
workflow.add_node("evidence_supported", evidence_supported_node)
workflow.add_node("uncertain", uncertain_node)
workflow.add_node("final_report", final_report_node)

workflow.add_edge(START, "interpret_prediction")
workflow.add_edge("interpret_prediction", "confidence_analysis")

workflow.add_conditional_edges(
    "confidence_analysis",
    confidence_router,
    {
        "supported": "evidence_supported",
        "uncertain": "uncertain"
    }
)

workflow.add_edge("evidence_supported", "final_report")
workflow.add_edge("uncertain", "final_report")
workflow.add_edge("final_report", END)

breast_agent = workflow.compile()

print("LangGraph breast-density explanation agent is ready.")
```

```python
# ============================================================
# SINGLE CELL: LangGraph + Hugging Face LLM Agent
# Breast Density Evidence-Guided Report Generation
# Run this AFTER Grad-CAM generation, where gradcam_df already exists
# ============================================================

!pip -q install langgraph huggingface_hub tqdm

import os
import time
import pandas as pd
from typing import TypedDict, Optional
from tqdm.auto import tqdm
from langgraph.graph import StateGraph, START, END
from huggingface_hub import InferenceClient

# ============================================================
# 1. Load Hugging Face token
# ============================================================

try:
    from google.colab import userdata
    HF_TOKEN = userdata.get("HF_TOKEN")
except Exception:
    HF_TOKEN = None

if HF_TOKEN is None:
    from getpass import getpass
    HF_TOKEN = getpass("Paste your Hugging Face token: ")

os.environ["HF_TOKEN"] = HF_TOKEN

print("HF token loaded:", HF_TOKEN is not None)

# ============================================================
# 2. Choose lightweight Hugging Face LLM
# ============================================================

# HF_LLM_MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
# HF_LLM_MODEL = "openai/gpt-oss-20b"
HF_LLM_MODEL = "meta-llama/Llama-3.1-8B-Instruct"

client = InferenceClient(
    model=HF_LLM_MODEL,
    token=os.environ["HF_TOKEN"]
)

print("Using Hugging Face LLM:", HF_LLM_MODEL)

# ============================================================
# 3. Check required dataframe
# ============================================================

required_columns = ["image_path", "gradcam_path", "predicted_class", "confidence"]

missing_cols = [c for c in required_columns if c not in gradcam_df.columns]

if missing_cols:
    raise ValueError(f"gradcam_df is missing required columns: {missing_cols}")

print("gradcam_df loaded with cases:", len(gradcam_df))

# ============================================================
# 4. LangGraph state
# ============================================================

class BreastDensityAgentState(TypedDict):
    image_path: str
    gradcam_path: str
    predicted_class: str
    confidence: float

    density_description: Optional[str]
    confidence_level: Optional[str]
    evidence_status: Optional[str]
    recommendation_level: Optional[str]
    llm_prompt: Optional[str]
    final_report: Optional[str]
    used_fallback: Optional[bool]


# ============================================================
# 5. Density class descriptions
# ============================================================

DENSITY_DESCRIPTIONS = {
    "A": "almost entirely fatty breast tissue",
    "B": "scattered fibroglandular breast density",
    "C": "heterogeneously dense breast tissue",
    "D": "extremely dense breast tissue"
}

# ============================================================
# 6. LangGraph nodes
# ============================================================

def interpret_prediction_node(state: BreastDensityAgentState):
    pred = state["predicted_class"]

    return {
        "density_description": DENSITY_DESCRIPTIONS.get(
            pred,
            "unknown breast density category"
        )
    }


def confidence_analysis_node(state: BreastDensityAgentState):
    conf = float(state["confidence"])

    if conf >= 0.85:
        confidence_level = "high"
    elif conf >= 0.70:
        confidence_level = "moderate"
    else:
        confidence_level = "low"

    return {
        "confidence_level": confidence_level
    }


def confidence_router(state: BreastDensityAgentState):
    if state["confidence_level"] == "low":
        return "uncertain"
    return "supported"


def evidence_supported_node(state: BreastDensityAgentState):
    if state["confidence_level"] == "high":
        evidence_status = (
            "Grad-CAM visual evidence was generated for the predicted class. "
            "The classifier confidence is high, but the Grad-CAM map is only an explanatory visualization, not proof of the prediction."
        )
        recommendation_level = "standard evidence-guided research report"
    else:
        evidence_status = (
            "Grad-CAM visual evidence was generated for the predicted class. "
            "The classifier confidence is moderate, so the result should be interpreted cautiously."
        )
        recommendation_level = "cautious evidence-guided research report"

    return {
        "evidence_status": evidence_status,
        "recommendation_level": recommendation_level
    }


def uncertain_node(state: BreastDensityAgentState):
    return {
        "evidence_status": (
            "Grad-CAM visual evidence was generated, but the classifier confidence is low. "
            "The prediction should be treated as uncertain."
        ),
        "recommendation_level": "low-confidence research warning"
    }


def build_llm_prompt_node(state: BreastDensityAgentState):
    image_name = os.path.basename(state["image_path"])
    gradcam_name = os.path.basename(state["gradcam_path"])

    prompt = f"""
You are a research-writing assistant. Your only task is to format the provided model outputs into a short report.

You are NOT allowed to interpret the medical image.
You are NOT allowed to interpret the Grad-CAM heatmap content.
You are NOT allowed to describe visible tissue, lesion, abnormality, cancer, disease, or patient outcome.
You are NOT allowed to recommend further imaging, investigation, diagnosis, treatment, or clinical action.

Allowed facts:
- Image name: {image_name}
- The model predicted breast density class: {state["predicted_class"]}
- The density class corresponds to: {state["density_description"]}
- The model confidence score is: {state["confidence"]:.3f}
- The confidence level is: {state["confidence_level"]}
- A Grad-CAM heatmap file was generated: {gradcam_name}
- Evidence status: {state["evidence_status"]}
- Agent decision: {state["recommendation_level"]}

Forbidden phrases and ideas:
- "the image contains"
- "the heatmap shows tissue"
- "regions of interest"
- "additional imaging"
- "further investigation"
- "patient"
- "diagnosis"
- "cancer"
- "malignant"
- "benign"
- "lesion"
- "abnormality"
- "treatment"
- "clinical action"
- "clinical recommendation"

Write exactly four short sections:

1. Prediction Summary
2. Visual Evidence
3. Confidence Assessment
4. Research Note

Rules:
- Keep each section 1 to 2 sentences only.
- Do not add information beyond the allowed facts.
- Do not claim that Grad-CAM proves the prediction.
- Do not say the confidence is high unless the confidence level is high.
- Do not say the confidence is low unless the confidence level is low.
- The output must be suitable for a research paper, not a clinical report.
""".strip()

    return {
        "llm_prompt": prompt
    }


# ============================================================
# 7. Safe deterministic fallback report
# ============================================================

def make_safe_fallback_report(state: BreastDensityAgentState):
    image_name = os.path.basename(state["image_path"])
    gradcam_name = os.path.basename(state["gradcam_path"])

    return f"""
1. Prediction Summary
The model classified {image_name} as breast density class {state["predicted_class"]}, corresponding to {state["density_description"]}. The confidence score was {state["confidence"]:.3f}.

2. Visual Evidence
A Grad-CAM heatmap was generated for the predicted class and saved as {gradcam_name}. The heatmap is used only as a visual explanation of the model decision process, not as proof of the prediction.

3. Confidence Assessment
The confidence level is {state["confidence_level"]}. Based on the agent decision, this case is reported as: {state["recommendation_level"]}.

4. Research Note
This output is intended for research and explainability support only. It is not a clinical diagnosis or medical recommendation.
""".strip()


def report_has_forbidden_claims(report):
    forbidden_terms = [
        "the image contains",
        "heatmap shows",
        "heatmap highlights tissue",
        "regions of interest",
        "additional imaging",
        "further investigation",
        "patient",
        "diagnosis",
        "cancer",
        "malignant",
        "benign",
        "lesion",
        "abnormality",
        "treatment",
        "clinical action",
        "clinical recommendation",
        "medical recommendation",
        "disease",
        "tumor",
        "carcinoma"
    ]

    report_lower = report.lower()

    for term in forbidden_terms:
        if term in report_lower:
            return True

    return False


# ============================================================
# 8. Hugging Face LLM call
# ============================================================

def call_hf_llm(prompt, max_retries=3):
    messages = [
        {
            "role": "system",
            "content": (
                "You are a strict research-report formatting assistant. "
                "You must not add clinical or visual interpretations beyond the provided facts."
            )
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    last_error = None

    for attempt in range(max_retries):
        try:
            response = client.chat_completion(
                messages=messages,
                max_tokens=220,
                temperature=0.0,
                top_p=0.8
            )

            return response.choices[0].message.content.strip()

        except Exception as e:
            last_error = e
            time.sleep(2)

    return f"LLM report generation failed. Error: {str(last_error)}"


def llm_report_node(state: BreastDensityAgentState):
    llm_report = call_hf_llm(state["llm_prompt"])

    used_fallback = False

    if report_has_forbidden_claims(llm_report):
        final_report = make_safe_fallback_report(state)
        used_fallback = True
    else:
        final_report = llm_report

    return {
        "final_report": final_report,
        "used_fallback": used_fallback
    }


# ============================================================
# 9. Build LangGraph workflow
# ============================================================

workflow = StateGraph(BreastDensityAgentState)

workflow.add_node("interpret_prediction", interpret_prediction_node)
workflow.add_node("confidence_analysis", confidence_analysis_node)
workflow.add_node("evidence_supported", evidence_supported_node)
workflow.add_node("uncertain", uncertain_node)
workflow.add_node("build_llm_prompt", build_llm_prompt_node)
workflow.add_node("llm_report", llm_report_node)

workflow.add_edge(START, "interpret_prediction")
workflow.add_edge("interpret_prediction", "confidence_analysis")

workflow.add_conditional_edges(
    "confidence_analysis",
    confidence_router,
    {
        "supported": "evidence_supported",
        "uncertain": "uncertain"
    }
)

workflow.add_edge("evidence_supported", "build_llm_prompt")
workflow.add_edge("uncertain", "build_llm_prompt")
workflow.add_edge("build_llm_prompt", "llm_report")
workflow.add_edge("llm_report", END)

breast_llm_agent = workflow.compile()

print("LangGraph + Hugging Face LLM agent compiled successfully.")

# ============================================================
# 10. Test on one image
# ============================================================

sample_row = gradcam_df.iloc[0]

test_input_state = {
    "image_path": sample_row["image_path"],
    "gradcam_path": sample_row["gradcam_path"],
    "predicted_class": sample_row["predicted_class"],
    "confidence": float(sample_row["confidence"]),

    "density_description": None,
    "confidence_level": None,
    "evidence_status": None,
    "recommendation_level": None,
    "llm_prompt": None,
    "final_report": None,
    "used_fallback": None
}

test_result = breast_llm_agent.invoke(test_input_state)

print("\n" + "=" * 70)
print("TEST REPORT FOR ONE IMAGE")
print("=" * 70)
print(test_result["final_report"])
print("\nUsed fallback:", test_result["used_fallback"])

# ============================================================
# 11. Run on all Grad-CAM outputs
# ============================================================

llm_agent_outputs = []

for row in tqdm(
    gradcam_df.itertuples(index=False),
    total=len(gradcam_df),
    desc="Generating LLM agent reports",
    unit="case"
):
    input_state = {
        "image_path": row.image_path,
        "gradcam_path": row.gradcam_path,
        "predicted_class": row.predicted_class,
        "confidence": float(row.confidence),

        "density_description": None,
        "confidence_level": None,
        "evidence_status": None,
        "recommendation_level": None,
        "llm_prompt": None,
        "final_report": None,
        "used_fallback": None
    }

    result = breast_llm_agent.invoke(input_state)

    llm_agent_outputs.append({
        "image_path": row.image_path,
        "gradcam_path": row.gradcam_path,
        "predicted_class": row.predicted_class,
        "confidence": row.confidence,
        "density_description": result["density_description"],
        "confidence_level": result["confidence_level"],
        "evidence_status": result["evidence_status"],
        "recommendation_level": result["recommendation_level"],
        "used_fallback": result["used_fallback"],
        "llm_prompt": result["llm_prompt"],
        "final_report": result["final_report"]
    })

llm_agent_df = pd.DataFrame(llm_agent_outputs)

display(llm_agent_df)

# ============================================================
# 12. Save results
# ============================================================

llm_agent_csv_path = "/content/breast_density_llm_agent_reports.csv"
llm_agent_df.to_csv(llm_agent_csv_path, index=False)

print("\nSaved LLM agent reports:", llm_agent_csv_path)

fallback_count = int(llm_agent_df["used_fallback"].sum())
total_count = len(llm_agent_df)

print(f"Fallback reports used: {fallback_count}/{total_count}")

print("\nPipeline completed:")
print("Breast image → pretrained classifier → Grad-CAM → LangGraph evidence routing → HF LLM report → guardrail fallback")
```
